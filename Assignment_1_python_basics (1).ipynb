```python
# 1. Simple ATM Calculator

def wdrl(curr_balance,wdrl_amt):
    new_balance=curr_balance-wdrl_amt
    if new_balance<0:
      print("!! Somthing Wrong !! , Insufficiant balance")
    else:
      print("Transaction sucessfull,Your new balance = ",new_balance)
      return new_balance
wdrl(1000,1500)
```

```python
# 2. Electricity Bill Generator

units=eval(input("Enter the nuber of units consumed: "))

if units <= 100:
  bill=units*5

elif units <= 200:
  bill=(100*5)+((units-100)*7)

else:
  bill=(100*5)+(100*7)+((units-200)*10)

print("Bill Amount =","Rs",bill)
```

```python
# 3. Temperature Converter

t=eval(input("Enter Temprature : "))
u=input("Enter Unit(F/C) : ")

if u == "C":
  f = (t * 9/5) + 32
  print("Temprature in Fahrenheit = ",f,"°F")
elif u == "F":
 c = (t - 32) * 5/9
 print("Temprature in Celsius = ",c,"°C")
else:
  print("Invalid Unit,Please enter C or F")
```

```python
# 4. Mobile Recharge Validity

from datetime import datetime

recharge = input("Enter the Recharge date(YYYY-MM-DD) : ")
today = input("Enter Today's date(YYYY-MM-DD) : ")

recharge_date = datetime.strptime(recharge, "%Y-%m-%d")
today_date = datetime.strptime(today, "%Y-%m-%d")

days = (today_date - recharge_date).days

if days < 28:
  print("Plan is still valid")
elif days == 28:
  print("Expiring soon,Recharge Immediately")
else:
  print("Plan expired,Need to Recharge")
```

```python
# 6. Sum of even numbers

x = eval(input("Enter the number: "))

def check(x,sum=0):

  for i in range(x,1,-1):
    if i%2==0:
      sum+=i
  print(f"Summ of even no from {x} to 1 is :",sum)

check(x)
```

```python
# 7. Dictionary – Student Marks Summary

students = {}

i = 0

while i < 3:
    name = input("Enter name: ")
    marks = int(input("Enter marks: "))

    students[name] = marks

    i += 1

average = sum(students.values()) / len(students)

print("Student Marks:", students)
print("Average Marks:", average)
```

```python
# 8. Dictionary – Word Frequency

sentence = input("Enter the sentence: ")
words = sentence.split()
frequency = {}

for i in words:
  if i in frequency:
    frequency[i] += 1
  else:
    frequency[i] = 1

for i in frequency:
  print(i,":",frequency[i])
```

```python
# 8. Functions – Check Prime

num = int(input("Enter the number"))
prime = lambda n: n>1 and all(n % i != 0 for i in range(2,n))

if prime(num):
  print(num,"is a prime number")
else:
  print(num,"is not a prime number")
```

```python
# 9. Vowel Counter

str = input("Enter the string: ")
vowl = ["a","e","i","o","u"]

count = len([i for i in str if i in vowl])

print("Number of vowels : ",count)
```
